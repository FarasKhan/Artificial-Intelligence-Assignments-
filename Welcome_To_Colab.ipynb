```

flights = [
    {"id": "TK401", "from": "Tokyo", "to": "Sydney", "price": 650},
    {"id": "TK402", "from": "Tokyo", "to": "Sydney", "price": 710},
    {"id": "SY501", "from": "Sydney", "to": "Singapore", "price": 420},
    {"id": "SG601", "from": "Singapore", "to": "Dubai", "price": 530},
    {"id": "DX701", "from": "Dubai", "to": "Tokyo", "price": 890},
]

# Get user input
origin = input("Origin city: ").strip().lower()
destination = input("Destination city: ").strip().lower()

# Filter matching flights
available_options = [
    item for item in flights
    if item["from"].lower() == origin and item["to"].lower() == destination
]

if available_options:
    print(f"\n--- Direct Flights Found ({len(available_options)}) ---")
    for flight in available_options:
        print(f"[{flight['id']}] {flight['from']} -> {flight['to']} | ${flight['price']}")

    choice = input("\nSelect Flight ID to reserve: ").strip().upper()

    # Confirm valid selection
    valid_ids = [f["id"] for f in available_options]
    if choice in valid_ids:
        print(f"Reservation confirmed for flight {choice}!")
    else:
        print("Invalid flight selection.")
else:
    print("\nNo direct flights match your route.")
```
